```python
import pandas as pd

troop_names = ['Barbarian', 'Archer', 'Giant', 'Wizard', 'Dragon']

hit_points = [125, 100, 700, 230, 1900]

damage_points = [20, 30, 50, 90, 140]

housing_space = [ 1, 1, 5, 4, 20]

training_cost = [100, 200, 1500, 3500, 25000]


rows = []

for i in range(5):

      troop = troop_names[i]
      hp = hit_points[i]
      dmg = damage_points[i]
      housing = housing_space[i]
      cost = training_cost[i]

      rows.append({

            'Troop': troop,
            'HP': hp,
            'Damage': dmg,
            'Housing': housing,
            'Training_Cost': cost,

        })

df = pd.DataFrame(rows)
df.to_csv('coc_troops.csv', index=False)
print(f'Generated {len(df)} rows-> coc_troops.csv')
```

```python
import plotly.express as px
import pandas as pd

df = pd.read_csv('coc_troops.csv')

fig = px.bar(

    df,

    x='Troop',
    y='Damage',

    title='Clash of Clans Damage Comparison',

    labels={
        'Troop': 'Troop Name',
        'Damage': 'Damage Points',
    },

    color='Troop',
    text= 'Damage'
)

fig.update_layout(

    xaxis_title='Troop Name',
    yaxis_title='Damage Points',

    showlegend=False,

    height=500,

    title_x=0.5
)

fig.show()
```

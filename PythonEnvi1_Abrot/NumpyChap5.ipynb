```python
# Exercise B

import numpy as np

array_ones = np.ones((5, 5), dtype = int)

print('5x5 Array:')
print(array_ones)

print('\nShape:')
print(array_ones.shape)

print('\nSize:')
print(array_ones.size)

print('\nNumber of Dimensions:')
print(array_ones.ndim)
```

### C.2 Feature Selection
**Exercise C.4 (filter).** Find features whose absolute correlation with an *earlier* feature exceeds 0.95.

```python
# Compute correlation matrix
corr_matrix = X.corr().abs()

# Get upper triangle of correlation matrix
upper_tri = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))

# Find features with correlation greater than 0.95 with an earlier feature
to_drop = [column for column in upper_tri.columns if any(upper_tri[column] > 0.95)]
print(f"Redundant features to drop (>0.95 correlation): {to_drop}")
```

## D. Evaluate Algorithm

We use **stratified 10-fold cross-validation** on the training set as our test harness. Scaling is placed inside a `Pipeline` so it is re-fitted in every fold: this avoids leakage.

**Exercise D.1.** Define the six candidate models: Logistic Regression, LDA, KNN, CART, Gaussian Naive Bayes and SVM.

```python
seed = 7
kfold = StratifiedKFold(n_splits=10, shuffle=True, random_state=seed)

# Define dictionary named models with keys LR, LDA, KNN, CART, NB, SVM
models = {
    "LR": LogisticRegression(random_state=seed),
    "LDA": LinearDiscriminantAnalysis(),
    "KNN": KNeighborsClassifier(),
    "CART": DecisionTreeClassifier(random_state=seed),
    "NB": GaussianNB(),
    "SVM": SVC(random_state=seed)
}

def spot_check(models, X_tr, scale):
    """Cross-validated accuracy for every model (optionally scaled inside the pipeline)."""
    results = {}
    for name, model in models.items():
        steps = ([("scale", StandardScaler())] if scale else []) + [("model", model)]
        results[name] = cross_val_score(Pipeline(steps), X_tr, y_train,
                                        cv=kfold, scoring="accuracy")
    return results
```

**Exercise D.2.** Run the spot-check on all features, once **raw** and once **scaled**, and tabulate the mean accuracies.

```python
# Call spot_check twice: once with scale=False, once with scale=True
results_raw = spot_check(models, X_train, scale=False)
results_scaled = spot_check(models, X_train, scale=True)

# Build a DataFrame of mean CV accuracy scores
summary = pd.DataFrame({
    "raw": {name: np.mean(scores) for name, scores in results_raw.items()},
    "scaled": {name: np.mean(scores) for name, scores in results_scaled.items()}
})

print("Baseline Spot-Check Results (Mean CV Accuracy):")
display(summary)
```

✏️ **Question D.3.** Which models benefit from scaling, and why?

> 1. **KNN (K-Nearest Neighbors):** Benefits significantly because it calculates Euclidean distance between instances. Features with larger absolute values would dominate the distance calculation without scaling.
> 2. **SVM (Support Vector Machines):** Benefits greatly because optimization margins and kernel computations (like RBF) depend heavily on the scale of the features. Large scales can cause numeric instability or skewed margins.
> 3. **LR (Logistic Regression) with L2 regularization:** Benefits because regularization penalizes the size of coefficients. Unscaled features would result in uneven regularization penalty application.
> 4. **LDA (Linear Discriminant Analysis):** Benefits by ensuring covariance estimations are numerically stable and balanced.
> 5. **Tree-based models (CART / Decision Trees):** Do not benefit from scaling because they make split decisions sequentially based on single-feature thresholds, which are scale-invariant.

### E.1 Algorithm Tuning
**Exercise E.1.** Tune the SVM (`C`, `gamma`, `kernel`) with `GridSearchCV` inside a scaling pipeline.

```python
# TODO: grid search over C=[0.1,1,10,100], gamma=["scale",0.001,0.01,0.1], kernel=["rbf","linear"]
...
```

```python
pipeline_svm = Pipeline([("scale", StandardScaler()), ("model", SVC(random_state=seed))])

param_grid_svm = {
    "model__C": [0.1, 1, 10, 100],
    "model__gamma": ["scale", 0.001, 0.01, 0.1],
    "model__kernel": ["rbf", "linear"]
}

grid_svm = GridSearchCV(pipeline_svm, param_grid=param_grid_svm, cv=kfold, scoring="accuracy", n_jobs=-1)
grid_svm.fit(X_train, y_train)

print(f"Best SVM CV Accuracy: {grid_svm.best_score_:.4f}")
print(f"Best SVM Parameters: {grid_svm.best_params_}")
```

**Exercise E.3.** Spot-check four ensemble methods (Random Forest, Extra Trees, Gradient Boosting, AdaBoost) and one voting ensemble of the tuned SVM, tuned KNN and Random Forest.

```python
# TODO: dictionary ensembles, score with spot_check(scale=False), then a hard VotingClassifier
...
```

```python
ensembles = {
    "RF": RandomForestClassifier(random_state=seed),
    "ET": ExtraTreesClassifier(random_state=seed),
    "GBM": GradientBoostingClassifier(random_state=seed),
    "AB": AdaBoostClassifier(random_state=seed)
}

# Spot-check the individual ensemble models (without internal scaling, as they are tree-based ensembles)
ensemble_results = spot_check(ensembles, X_train, scale=False)

# Instantiate tuned estimators for the Voting ensemble
tuned_svm = SVC(C=grid_svm.best_params_["model__C"],
                gamma=grid_svm.best_params_["model__gamma"],
                kernel=grid_svm.best_params_["model__kernel"],
                random_state=seed)

tuned_knn = KNeighborsClassifier(n_neighbors=grid_knn.best_params_["model__n_neighbors"],
                                 weights=grid_knn.best_params_["model__weights"])

rf_model = RandomForestClassifier(random_state=seed)

voting_clf = VotingClassifier(estimators=[
    ("svm", Pipeline([("scale", StandardScaler()), ("model", tuned_svm)])),
    ("knn", Pipeline([("scale", StandardScaler()), ("model", tuned_knn)])),
    ("rf", rf_model)
], voting="hard")

voting_scores = cross_val_score(voting_clf, X_train, y_train, cv=kfold, scoring="accuracy")

ensemble_results["Voting"] = voting_scores
print("Ensembles Mean CV Accuracy:")
for name, scores in ensemble_results.items():
    print(f"  {name}: {np.mean(scores):.4f}")
```

1. **Leakage.** Feature selection in Part C used the whole dataset. Explain why this can inflate CV scores and how you would fix it using a `Pipeline`.

- Performing feature selection on the entire dataset before splitting it into cross-validation folds leaks information about the validation sets into the training process. The model essentially "sees" characteristics of the validation folds beforehand, which artificially inflates cross-validation performance. To fix this, we should encapsulate feature selection (e.g., `SelectKBest`) inside a scikit-learn `Pipeline` alongside the scaler and estimator so that the feature selection step is fitted exclusively on the active training folds during each cross-validation run.

2. **Metric choice.** Which metric would you report to a clinician, and why is accuracy alone insufficient?

- I would report Recall (Sensitivity) for the malignant class to a clinician, along with False Negatives and ROC-AUC. Accuracy alone is insufficient because it treats all classification errors equally. In breast cancer diagnosis, a false negative (falsely diagnosing a malignant tumor as benign) has catastrophic clinical consequences (delayed treatment), whereas a false positive leads only to further testing. Thus, focusing on maximizing recall ensures we catch as many true malignant cases as possible.

I would also inspect the variance of the scores across cross-validation folds; a model with slightly lower accuracy but much lower variance is generally more robust and preferable.

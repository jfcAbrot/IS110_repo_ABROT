```python
# EXERCISE 1:Rewrite your pay computation with time-and-a-half for overtime and create a function called compute pay which takes two parameters (hours and rate).

def compute_pay(hours, rate):
  overtime = max(0, hours - 40)
  regular = hours - overtime
  return regular * rate + overtime * rate * 1.5

while True:
  try:
    hours = float (input('Enter hours: '))
    rate = float (input('Enter rate: '))

    if hours > 0 and rate < 0:
      print('Hours and rate must be non-negative')
      continue

    break
  except ValueError:
    print('Invalid input, please enter numeric values')

pay = compute_pay(hours, rate)
print(f'Pay: {pay:.2f}')
```
